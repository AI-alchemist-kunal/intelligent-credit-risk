```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", 120)
pd.set_option("display.float_format", "{:.3f}".format)

DATA_PATH = Path("../data/raw/application_train.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
df.head()
```

```python

df.info()

print("\nNumerical features:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical features:")
print(df.select_dtypes(include="object").columns.tolist())

print("\nDuplicate application IDs:")
print(df["SK_ID_CURR"].duplicated().sum())
```

```python
## Analysis of the target variable
target_counts = df["TARGET"].value_counts()
target_percentage = df["TARGET"].value_counts(
    normalize=True
) * 100

print("Counts:")
print(target_counts)

print("\nPercentages:")
print(target_percentage)

sns.countplot(data=df, x="TARGET")
plt.title("Repayment Difficulty Distribution")
plt.xlabel("TARGET")
plt.ylabel("Number of Applications")
plt.show()
```

```python
## Investigate missing values
missing_pct = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_summary = missing_pct.reset_index()
missing_summary.columns = [
    "feature",
    "missing_percentage"
]

display(missing_summary.head(20))

plt.figure(figsize=(10, 6))
sns.barplot(
    data=missing_summary.head(20),
    x="missing_percentage",
    y="feature"
)
plt.title("Top 20 Features by Missing Percentage")
plt.xlabel("Missing Values (%)")
plt.tight_layout()
plt.show()
```

```python
#@ Exploratory analysis of borrower numerical features
features = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "DAYS_BIRTH",
    "DAYS_EMPLOYED"
]

display(df[features].describe())

print(
    df.groupby("TARGET")[
        ["AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY"]
    ].median()
)
```

```python

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="TARGET",
    y="AMT_CREDIT",
    showfliers=False
)

plt.title("Loan Amount by Repayment Difficulty")
plt.xlabel("TARGET")
plt.ylabel("Credit Amount")
plt.show()
```
